```python
# Initialize a Wikipedia retriever to fetch top 3 articles in English

from langchain_community.retrievers import WikipediaRetriever

retriever = WikipediaRetriever(top_k_results=3, lang='en')
```

```python
# Retrieve Wikipedia articles related to "Michael Jackson" and print the count

query = "Michael Jackson"
docs = retriever.invoke(query)
print("Number of Documents retrieved : ", len(docs))
```

```python
# Print each retrieved Wikipedia article with numbered headers and content

for i, doc in enumerate(docs):
    print(f"\n--------- Result {i+1} ---------")
    print(f"Content : {doc.page_content}")
```
