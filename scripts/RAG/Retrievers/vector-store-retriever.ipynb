```python
# Create a vector store from movie documents using HuggingFace embeddings for semantic search

docs = [doc1, doc2, doc3, doc4, doc5]  # List of movie documents
embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")  # Load embedding model

vector_store = Chroma.from_documents(  # Create Chroma vector store from documents
    documents=docs,
    embedding=embedding_model,
    collection_name="my_collection"
)
```

```python
# Convert the vector store into a retriever to fetch top 2 semantically similar documents

retriever = vector_store.as_retriever(search_kwargs={"k": 2})
```

```python
# Search the vector store for documents relevant to the query "Movie about space"

query = "Movie about space"
results = retriever.invoke(query)  # Retrieve top matching documents
```

```python
# Print the content of each retrieved document with a numbered header
for i, doc in enumerate(results):
    print(f"\n--------- Result {i+1} ---------")
    print(doc.page_content)
```
